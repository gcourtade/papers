## Figure martini

```python
data = np.loadtxt('COLVAR_opes')
```

```python
data1 = np.loadtxt('COLVAR1_opes')
```

```python
KB = 0.0083144621
T0 = 300
KBT = KB*T0
d = data[:,1]
ang = data[:,2]
bias1 = data[:, 3]
bias2 = data1[:,2]
bias = bias1 + bias2
logweights = bias/KBT
```

```python
grid_d   = np.linspace(1.0, 9.0, 500)
grid_ang = np.linspace(-1,1, 500)
X, Y = np.meshgrid(grid_d, grid_ang, indexing='ij')
bandwidth = np.array([0.15, 0.15])
```

```python
fes_2d = calcFES((X, Y), bandwidth, (d, ang),
                 logweights=logweights,
                 mintozero=True)
```

```python
vmin, vmax = 0, 15 


# Creare un grafico a contorno utilizzando le estetiche di Seaborn
plt.figure(figsize=(10, 8))
contour = plt.contourf(X, Y, fes_2d/ 4.184, levels=20, cmap="magma", vmin=vmin, vmax=vmax)
contour_lines = plt.contour(X, Y, fes_2d/ 4.184, levels=20, colors="white", linewidths=0.5)
plt.colorbar(contour, label="Free Energy (kcal/mol)")
plt.clabel(contour_lines, fmt='%1.1f')

# Aggiungere titolo e etichette con lo stile Seaborn

plt.xlabel(r"$d_A$ (nm)")
plt.ylabel(r"$cos(\tau)$")
plt.savefig("fes.svg", bbox_inches="tight", dpi=600)
plt.show()
```

```python
def triple_exp(t, k1, t1, k2, t2, k3, t3):
    return k1 * np.exp(-t / t1) + k2 * np.exp(-t / t2) + k3 * np.exp(-t / t3) + 1
```

```python
p0 = [1, 1, 1, 1, 1,5]
t = df['Block_ns']
y = 1/df['Norm_omega']
```

```python
params, covariance = curve_fit(triple_exp, t, y, p0=p0)
```

```python
k1, t1, k2, t2, k3, t3 = params
print(f"k1={k1:.3f}, t1={t1:.3f}, k2={k2:.3f}, t2={t2:.3f}, k3={k3:.3f}, t3={t3:.3f}")
```

```python
t_fit = np.linspace(10, 4000, 200)
y_fit = triple_exp(t_fit, *params)

plt.scatter(t, y, label='Data', color='blue')
plt.plot(t_fit, y_fit, label='Exponent Fitting', color='red')
plt.xlabel('time (ns)')
plt.ylim(0.8,2.0)
plt.ylabel('Bootstrapped/Blocked')
plt.legend()
plt.savefig("fitting.png", dpi=600)
plt.show()
```

```python
# Creazione del jointplot con KDE
g = sns.jointplot(
    data=df_glc_rep1,
    x='phi',
    y='theta',
    kind="kde",
    fill=True,
    cmap="Greens",
    alpha=1,
    height=5,      # Altezza complessiva in pollici
    aspect=1.0     # Rapporto width/height
)

g.fig.set_size_inches(figsize)

# Definizione dei limiti per la scala dei colori
vmin, vmax = 0.0, 1.4

# Aggiungiamo il grafico a dispersione sul joint axis
scatter = g.ax_joint.scatter(
    df_glc_rep1['phi'],
    df_glc_rep1['theta'],
    c=df_glc_rep1['CV'],
    cmap='magma',
    edgecolor='k',
    s=20,
    vmin=vmin,
    vmax=vmax,
    linewidth=0.4
)

# Aggiustiamo il layout per lasciare spazio a destra per la colorbar
g.fig.subplots_adjust(right=0.85)

# Aggiungiamo la barra dei colori nella zona libera a destra del plot principale
# [left, bottom, width, height] in coordinate relative alla figura
cbar_ax = g.fig.add_axes([0.88, 0.15, 0.03, 0.7])
cbar = g.fig.colorbar(scatter, cax=cbar_ax)
cbar.set_label(r'$d_k$ (nm)')

# Impostiamo le etichette con font leggibile e unità tra parentesi
g.set_axis_labels(r"$\gamma$ (°)", r"$\theta$ (°)")

# Impostazione dei limiti degli assi
g.ax_joint.set_xlim(-300, 300)
g.ax_joint.set_ylim(-10, 200)

# Personalizzazione dei tick e rimozione delle spine superiori e destre
#g.ax_joint.tick_params(direction="in", labelsize=9)
#g.ax_joint.spines["top"].set_visible(False)
#g.ax_joint.spines["right"].set_visible(False)
plt.savefig("angles_glc_rep1.svg", format="svg", bbox_inches='tight')
plt.show()
```
