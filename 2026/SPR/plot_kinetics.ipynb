```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
files = ['M1_fit_solution.txt', 'M2_fit_solution.txt', 'M3_fit_solution.txt', 'M4_fit_solution.txt']
folders = ['PET_kinetics', 'CellulosePET_kinetics']
order = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']
fig, ax = plt.subplot_mosaic([['a', 'e'],
                              ['b', 'f'],
                              ['c', 'g'],
                              ['d', 'h']], figsize=(14, 10), sharex=True)
i = 0
for folder in folders:
    for file in files:
        a = order[i]
        print(a)
        data = np.loadtxt(f'{folder}/{file}', delimiter='\t', skiprows=1)
        x = data[:,0]
        y = data[:,1]
        GE = data[:,2]
        GI = data[:,3]
        GT = data[:,4]
        ax[a].plot(x, y, 'o', markersize=4, color='tab:blue', alpha=0.5, label='Experimental data')
        ax[a].plot(x, GE, 'k--', linewidth=2, label=r'Reversibly bound ($\Gamma_E$)')
        ax[a].plot(x, GI, 'k:', linewidth=2, label=r'Irreversibly bound ($\Gamma_I$)')
        ax[a].plot(x, GT, 'k-', linewidth=2, label=r'Total bound ($\Gamma_T$)')
    
       # plt.axvline(x=transition_time_fitted, color='k', linestyle='--',
        #            label=f'Washoff at {transition_time_fitted:.1f} min')
        
        if a in ['d', 'h']:
            ax[a].set_xlabel('Time (s)', fontsize=14)




        i+=1
#plt.xlim(min(x_data), max(x_data))
#plt.ylim(0, 1.2)
#plt.title('Decomposition of Enzyme Binding into Reversible and Irreversible Components', fontsize=16)
fig.text(0.09, 0.5, 'Normalized signal', ha='center', va='center', rotation='vertical', fontsize=14)
fig.text(0.51, 0.5, 'Normalized signal', ha='center', va='center', rotation='vertical', fontsize=14)

ax['a'].legend(fontsize=8, loc='lower right', labelspacing=0, frameon=True, framealpha=0.5)
ax['a'].set_title('PET', fontsize=14)
ax['e'].set_title('Cellulose-PET', fontsize=14)

plt.savefig(f'all_fits.png', dpi=300)
#plt.grid(True)
plt.show()
```
