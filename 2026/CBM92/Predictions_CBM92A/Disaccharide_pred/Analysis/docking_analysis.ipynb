```python
# Save figures of affinity vs score

plt.rcParams.update({
    "font.size": 8,        # base font size
    "axes.labelsize": 8,   # axis labels
    "xtick.labelsize": 7,  # x tick labels
    "ytick.labelsize": 7,  # y tick labels
    "legend.fontsize": 7,  # legend
})

def plot_score(df):
    plt.figure(figsize=(2,2))

    # mapping end-value → marker
    markers = {0: 'x', 1: 'o', 2: '^'}

    # Loop through each timepoint
    for t, color in zip([48, 90, 132], ['r', 'k', 'b']):
        score_col = f"score_{t}"
        aff_col   = f"affinity_{t}"
        end_col   = f"end_{t}"

        # Plot each category separately so each gets its own marker
        for end_value in [0, 1, 2]:
            sub = df[df[end_col] == end_value]
            if len(sub) == 0:
                continue

            plt.plot(
                sub[score_col],
                sub[aff_col],
                linestyle='',
                marker=markers[end_value],
                alpha=0.5,
                markersize=3,
                color=color,
                label=f"W{t} (end={end_value})"
            )
    plt.xlabel('CNN score')
    plt.ylabel('CNN affinity')
    plt.xlim(0,1)
    plt.ylim(2.2,4)
    #plt.legend()
    
dfs = [aCel, bCel, aLam, bLam, aSoph, bSoph, aGent, bGent]
names = ['aCellobiose', 'bCellobiose', 'aLaminaribiose', 'bLaminaribiose', 'aSophorose', 'bSophorose', 'aGentiobiose', 'bGentiobiose']

for df, name in zip(dfs, names):
    plot_score(df)
    plt.title(f'{name}')
    plt.savefig(f'Plots/{name}_plot.png', dpi = 300, bbox_inches='tight')
```

```python
#Count non-reducing end dockings

def count_end_percent(df_list):
    results = {}

    for i, df in enumerate(df_list, start=1):
        df_name = f"df_{i}"
        results[df_name] = {}
        
        # Find all columns that start with "end_"
        end_cols = [col for col in df.columns if col.startswith('end_')]
        
        for col in end_cols:
            # Extract only values 1 or 2
            mask = df[col].isin([1, 2])
            valid = df.loc[mask, col]

            count_2 = (valid == 2).sum()
            count_1 = (valid == 1).sum()
            total = len(valid)

            percent_2 = 100 * count_2 / total if total > 0 else None
            
            results[df_name][col] = {
                "count_1": count_1,
                "count_2": count_2,
                "percent_2": percent_2
            }

    return results


dfs = [aCel, bCel, aLam, bLam, aSoph, bSoph, aGent, bGent]
count_end_percent(dfs)
```

```python
def overall_end_percent(df_list):
    combined_results = {}

    # Determine all end_* columns across the dfs
    all_end_cols = sorted(
        set(col for df in df_list for col in df.columns if col.startswith("end_"))
    )

    for col in all_end_cols:
        total_1 = 0
        total_2 = 0

        # Aggregate counts across all dataframes
        for df in df_list:
            if col not in df.columns:
                continue

            mask = df[col].isin([1, 2])  # exclude 0
            values = df.loc[mask, col]

            total_1 += (values == 1).sum()
            total_2 += (values == 2).sum()

        total = total_1 + total_2
        percent_2 = 100 * total_2 / total if total > 0 else None

        combined_results[col] = {
            "total_1": total_1,
            "total_2": total_2,
            "overall_percent_2": percent_2
        }

    return combined_results
```

## CNN Affinity


```python
def filter_affinities(df, scorelimit):
    df_48 = df[df['score_48'] >= scorelimit][['Unnamed: 0', 'score_48', 'affinity_48']]
    df_90 = df[df['score_90'] >= scorelimit][['Unnamed: 0', 'score_90', 'affinity_90']]
    df_132 = df[df['score_132'] >= scorelimit][['Unnamed: 0', 'score_132', 'affinity_132']]

    df_48['Kd'] = 10**(-df_48['affinity_48'])*1000   
    df_90['Kd'] = 10**(-df_90['affinity_90'])*1000 
    df_132['Kd'] = 10**(-df_132['affinity_132'])*1000 

    print('Trp48: Kd =', "{:.3f}".format(df_48['Kd'].mean()), '+/-', "{:.3f}".format(df_48['Kd'].std()), 'mM (based on', len(df_48), 'dockings).')
    print('Trp90: Kd =', "{:.3f}".format(df_90['Kd'].mean()), '+/-', "{:.3f}".format(df_90['Kd'].std()), 'mM (based on', len(df_90), 'dockings).')
    print('Trp132: Kd =', "{:.3f}".format(df_132['Kd'].mean()), '+/-', "{:.3f}".format(df_132['Kd'].std()), 'mM (based on', len(df_132), 'dockings).')

    Kds = pd.DataFrame()
    Kds['Res'] = ['W48', 'W90', 'W132']
    Kds['Kd'] = [df_48['Kd'].mean(), df_90['Kd'].mean(), df_132['Kd'].mean()]
    Kds['Std'] = [df_48['Kd'].std(), df_90['Kd'].std(), df_132['Kd'].std()]

    return [df_48, df_90, df_132], Kds

```

```python
lim = 0.85

print('\u03B1-Glucose:')
aGlc_dfs, aGlc_Kd = filter_affinities(aGlc, lim)
print('\u03B2-Glucose:')
bGlc_dfs, bGlc_Kd = filter_affinities(bGlc, lim)
print('\u03B1-Sophorose:') #B-1,2
aSoph_dfs, aSoph_Kd = filter_affinities(aSoph, lim)
print('\u03B2-Sophorose:')
bSoph_dfs, bSoph_Kd = filter_affinities(bSoph, lim)

print('\u03B1-Laminaribiose:') #B-1,3
aLam_dfs, aLam_Kd = filter_affinities(aLam, lim)
print('\u03B2-Laminaribiose:')
bLam_dfs, bLam_Kd = filter_affinities(bLam, lim)

print('\u03B1-Cellobiose:') #B-1,4
aCel_dfs, aCel_Kd = filter_affinities(aCel, lim)
print('\u03B2-Cellobiose:')
bCel_dfs, bCel_Kd = filter_affinities(bCel, lim)

print('\u03B1-Gentiobiose:') #B-1,6
aGent_dfs, aGent_Kd = filter_affinities(aGent, lim)
print('\u03B2-Gentiobiose:')
bGent_dfs, bGent_Kd = filter_affinities(bGent, lim)
```

```python
dfs = aGlc_Kd, bGlc_Kd, aSoph_Kd, bSoph_Kd, aLam_Kd, bLam_Kd, aCel_Kd, bCel_Kd, aGent_Kd, bGent_Kd 
labels = ['\u03B1-Glucose', '\u03B2-Glucose', '\u03B1-Sophorose', '\u03B2-Sophorose', '\u03B1-Laminaribiose', '\u03B2-Laminaribiose', '\u03B1-Cellobiose', '\u03B2-Cellobiose', '\u03B1-Gentiobiose', '\u03B2-Gentiobiose']

plt.figure(figsize = (3,3))
for i, df in enumerate(dfs):
    plt.errorbar(df['Res'], df['Kd'], yerr = df['Std'], fmt = 'o', elinewidth = 0.8, alpha=0.5, capsize = 3, label = labels[i])
plt.legend(fontsize=6,loc='center left', bbox_to_anchor=(1, 0.5))

plt.ylabel('$K_d$ (mM)')
#plt.title()
plt.savefig('Plots/Kd_bindingsites_boltz.png', dpi=600, bbox_inches = 'tight')
plt.savefig('Plots/Kd_bindingsites_boltz.pdf', bbox_inches = 'tight')
```
