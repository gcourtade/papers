```python
from Bio import AlignIO
```

```python
alignment_file = "clustalo-I20260421-225231-0679-51450131-p2m.aln-clustal_num"  

alignment = AlignIO.read(alignment_file, "clustal")

seqA = alignment[0].seq
seqB = alignment[1].seq

print("Alignment length:", alignment.get_alignment_length())
```

```python
for record in alignment:
    print(f"{record.id:15} {record.seq}")
```

```python
conserved_positions = []

for i in range(alignment.get_alignment_length()):
    a = seqA[i]
    b = seqB[i]

    if a == b and a != "-":
        conserved_positions.append(i)

print("Conserved alignment columns:", len(conserved_positions))
print(conserved_positions)
```

```python
def map_alignment_to_resi(seq):
    mapping = {}
    resi = 0

    for i, aa in enumerate(seq):
        if aa != "-":
            resi += 1
            mapping[i] = resi

    return mapping

mapA = map_alignment_to_resi(seqA)
mapB = map_alignment_to_resi(seqB)

resiA = [mapA[i] for i in conserved_positions if i in mapA]
resiB = [mapB[i] for i in conserved_positions if i in mapB]

print("Residues in BaCBM2:", resiA[:])
print("Residues in ScCBM2:", resiB[:])
```

```python
def format_resi(resi_list):
    return "+".join(map(str, resi_list))

pymol_script = f"""
# --- conserved residues ---
select conservedA, BaCBM2 and resi {format_resi(resiA)}
select conservedB, ScCBM2 and resi {format_resi(resiB)}

# --- show & color ---
show sticks, conservedA or conservedB
color yellow, conservedA or conservedB

set_color the_blue, [0.337, 0.706, 0.914] 
set_color the_orange, [0.902, 0.624, 0.000] 

color the_blue, BaCBM2
color the_orange, ScCBM2
show cartoon

"""

print(pymol_script)
```
