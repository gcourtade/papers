```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MultipleLocator
```

```python
filepath = 'Avicel_Rietveldt.txt'
df = pd.read_csv(filepath, quotechar="'")
df
```

```python
x = df.iloc[:, 0]
y_raw = df.iloc[:, 1]

plt.plot(x, y_raw, 'k', linewidth = 0.8, label = 'Raw data')
plt.plot(x, df['Ycalc'], 'r', linewidth = 0.8, alpha = 0.8, label = 'Refined data')
plt.plot(x, df['Bkg'], 'gray', linewidth = 0.8, alpha = 0.8, label = 'Background')
plt.plot(x, df['Diff'], 'darkgray', linewidth = 0.8, alpha = 0.8, label = 'Difference')
plt.plot(x, df['Cellulose_Nishiyama'], 'b', linewidth = 0.8, alpha = 0.8, label = 'Calc cellulose Iβ')
plt.plot(x.iloc[:360], df.iloc[:360]['Amorphous'], 'b--', linewidth = 0.8, alpha = 0.8, label = 'Calc amorphous')
plt.plot(x, df['Amorphous.1'], 'b--', linewidth = 0.8, alpha = 0.8, label = 'Calc amorphous')

plt.xlabel('2θ')
plt.ylabel('Intensity (arb. u)')
plt.title('Avicel')
plt.legend()

plt.savefig('Avicel_Rietveldt.svg', bbox_inches = 'tight')
```
