```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MultipleLocator
```

```python
filepath = 'CNC.txt'
df = pd.read_csv(filepath, sep = ';')
df = df.iloc[2:]
df = df.apply(pd.to_numeric, errors="coerce")

cols = ["Profile 1", "Profile 2", "Profile 3"] 
df[cols] = df[cols] * 1e6 #um

cols = ["Profile 1.1", "Profile 2.1", "Profile 3.1"] 
df[cols] = df[cols] * 1e9 #nm


df



```

```python
plt.figure(figsize = (3, 2.5))
plt.plot(df['Profile 1'], df['Profile 1.1'], linewidth = 0.8, label = '1', color = 'k')
plt.plot(df['Profile 2'], df['Profile 2.1'], linewidth = 0.8, label = '2', color = 'gray')
plt.plot(df['Profile 3'], df['Profile 3.1'], linewidth = 0.8, label = '3', color = 'darkblue')

plt.xlabel('Position (μm)')
plt.ylabel('Height (nm)')
plt.legend()
plt.savefig('CNC_profiles.svg')
```
